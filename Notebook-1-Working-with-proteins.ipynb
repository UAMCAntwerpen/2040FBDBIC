## 1. Introduction: structure-based design of MDM2–p53 PPI inhibitors

The p53 tumor suppressor controls cell-cycle arrest, DNA repair, senescence, and apoptosis. In many tumors, p53 is not mutated but is functionally silenced by MDM2, an E3 ubiquitin ligase that binds p53, blocks its transcriptional activity, and promotes p53 ubiquitination and degradation. This makes the MDM2–p53 protein–protein interaction an attractive pharmacological target, especially in tumors with wild-type TP53 and MDM2 overexpression or amplification.  

Structure-based drug design focuses on the N-terminal p53-binding pocket of MDM2. The p53 transactivation domain binds MDM2 mainly through three hydrophobic residues — commonly described as Phe19, Trp23, and Leu26 — which insert into complementary subpockets on MDM2. Small molecules such as nutlins, spiro-oxindoles, idasanutlin, milademetan, siremadlin, and navtemadlin were designed to mimic these p53 “hot-spot” interactions and competitively displace p53 from MDM2. Recent virtual-screening work still defines the design problem as finding MDM2 inhibitors at the p53-binding site.  

Pharmacologically, these compounds are p53 reactivators, not direct cytotoxics in the classical sense. By preventing MDM2 from binding p53, they stabilize p53 and restore transcription of p53 target genes such as CDKN1A/p21, PUMA, NOXA, and BAX, leading to growth arrest or apoptosis in cancer cells retaining functional p53. Their clinical use is therefore expected to depend strongly on biomarkers: TP53 wild-type status, MDM2 amplification/overexpression, intact downstream apoptotic signaling, and manageable toxicity in normal p53-competent tissues.

Clinically, MDM2 inhibitors have been explored in dedifferentiated liposarcoma, AML/MDS, myelofibrosis, CLL, glioblastoma, and other TP53-wild-type tumors. Development has been challenging because p53 activation can also affect normal tissues, causing dose-limiting toxicities such as thrombocytopenia, neutropenia, anemia, and gastrointestinal effects. Some trials have been disappointing: for example, phase III milademetan did not significantly improve progression-free survival over trabectedin in dedifferentiated liposarcoma. However, other agents remain clinically active areas of study, including navtemadlin in JAK-inhibitor–refractory myelofibrosis, where phase III BOREAS data showed improved spleen and symptom responses versus best available therapy. As of recent reviews, no MDM2 inhibitor has yet become a broadly established FDA-approved cancer therapy.

PPI should preferably be designed to bind MDM2, specifically the p53-binding cleft of MDM2, thereby blocking the MDM2–p53 interaction. Designing small molecules to bind p53 itself is less attractive for this mechanism because p53’s MDM2-binding segment is relatively shallow and intrinsically disordered before binding, whereas MDM2 provides a defined hydrophobic pocket suitable for structure-based design. Thus, these are best described as MDM2 antagonists that reactivate p53, rather than p53-binding inhibitors.

## 2. Working with proteins in the computer - the MDM2-p53 complex

For many proteins, the three-dimensional structure has been resolved using a technique termed <a href="https://www.youtube.com/watch?v=LioYrxinn28" target="_blank">X-ray crystallography</a>. All these structures are stored in a database called the <a href="https://www.rcsb.org" target="_blank">PDB Protein Data Bank</a>. You can search this database for your own protein and specify a large number of conditions, such as the organism, the presence of inhibitors, quality of the structure and so forth.

Protein crystal structures are evaluated using a combination of data quality, model quality, and geometric validation criteria. Key indicators include the **resolution** (level of structural detail), the **R-factor** (R<sub>work</sub>) and R<sub>free</sub> (agreement between observed and calculated diffraction data), and **B-factors** (atomic displacement or disorder). The Protein Data Bank (PDB) stores most of these metrics alongside each deposited structure, including resolution, R<sub>work</sub>/R<sub>free</sub>, crystallographic statistics (e.g., space group, unit cell), atomic coordinates with B-factors, and validation reports summarizing geometric quality and outliers. Each PDB protein structure is identified with a **unique 4-character code**, such as 1YCR.

#### 2.1 Resolution

In X-ray crystallography, **resolution** describes the level of detail visible in a protein structure and is directly related to the highest diffraction angles measured in the experiment. It is typically reported in Ångströms (Å) and reflects the smallest distance between two features in the electron density that can still be distinguished as separate. Physically, resolution comes from Bragg’s law, where higher-angle reflections correspond to finer structural detail. A lower numerical value means higher resolution:

* 1.0–1.5 Å → very high resolution (individual atoms clearly resolved, precise side-chain conformations)
* 1.8–2.5 Å → good/typical resolution (reliable backbone and most side chains)
* 2.5–3.5 Å → moderate resolution (overall fold clear, but less certainty in side chains)
* &gt;3.5 Å → low resolution (limited detail, often only overall architecture)

In practical terms, resolution determines how much confidence you can place in the atomic positions. At high resolution, you can model subtle features like hydrogen bonding networks or ligand interactions accurately—crucial for structure-based drug design. At lower resolution, interpretations become more approximate, and model bias can increase.

The reported resolution is a global measure, but real structures often have local variation in quality, meaning some regions (e.g., flexible loops) may be less well defined than others even in a high-resolution structure.

#### 2.2 R-factor

The **R-factor** (also called the R<sub>work</sub>) is a central measure of how well a crystallographic model explains the experimental diffraction data in X-ray crystallography. It quantifies the disagreement between the observed structure factor amplitudes from the experiment and the calculated amplitudes derived from the atomic model.

Formally, it is defined as:

$$R = \frac{\sum \left| F_{\text{obs}} - F_{\text{calc}} \right|}{\sum \left| F_{\text{obs}} \right|}$$

where $F_{\text{obs}}$ are the experimentally measured structure factor amplitudes and $F_{\text{calc}}$ are those predicted by the model. The sum runs over all reflections used in refinement.

In simple terms, the R-factor tells you how closely the model reproduces the experimental data:

* A lower R-factor means better agreement (typically 0.15–0.25 for a good protein crystal structure, depending on resolution).
* A higher R-factor indicates a poorer fit, suggesting errors in the model or insufficient data quality.

However, relying only on the R-factor can be misleading because the model is refined directly against the same data used to calculate it. To address this, crystallographers also use R<sub>free</sub>, which is calculated in the same way but on a small subset of reflections that were not used during refinement. This provides a more unbiased estimate of model quality and helps detect overfitting.

#### 2.3 B-factors

**B-factors** (also called temperature factors or atomic displacement parameters) describe how much an atom appears to “move” or be disordered within a crystal structure. In X-ray crystallography, atoms are not static—they may vibrate, adopt multiple conformations, or be poorly defined in the electron density. The B-factor captures this as a measure of how spread out the electron density is around a given atomic position.

Mathematically, the B-factor is related to the attenuation of diffraction intensity and can be expressed (in a simplified form) as:

$$B = 8\pi^2 \langle u^2 \rangle$$

where $\langle u^2 \rangle$ represents the mean square displacement of the atom from its average position.

In practice:

* Low B-factors (e.g., 10–30 Å²) indicate well-ordered atoms with clear electron density and high confidence in their positions.
* High B-factors (e.g., 50–80 Å²) suggest flexibility, disorder, or uncertainty—common in surface loops or poorly resolved regions.
* Extremely high values may indicate that the atom is not well supported by the data at all.

It’s important to note that B-factors reflect a combination of true atomic motion and experimental limitations (such as resolution or model inaccuracies). They are therefore best interpreted relatively within the same structure, rather than compared directly across different structures.

Since we are interested in identifying inhibitors of the MDM2-p53 interaction, so it may be of interest to look at structures of the MDM2-p53 complex, as well as structures of MDM2 alone in complex with an inhibitors.

The first high-resolution cocrystal structure of the human MDM2–p53 complex was solved by Kussie et al. in 1996 (PDB ID: <a href="https://www.rcsb.org/structure/1YCR" target="_blank">1YCR</a>). This structure has a resolution of 2.6 Å, so it is a suitable protein structure to work with. Subsequent work has led to the discovery of two additional structures of the MDM-p53 system, in particular the complexes between MDM4 and p53 (<a href="https://www.rcsb.org/structure/3DAB" target="_blank">3DAB</a> and <a href="https://www.rcsb.org/structure/3DAC" target="_blank">3DAC</a>).

````python
# Install the required code for visualisation in a notebook
!pip install py3Dmol
````

````python
import py3Dmol
import requests
# Example: load MDM2-p53 structure
pdb_id = "1YCR"
pdb_text = requests.get(f"https://files.rcsb.org/download/{pdb_id}.pdb").text
view = py3Dmol.view(width=800, height=500)
view.addModel(pdb_text, "pdb")

# Protein cartoon
view.setStyle({"hetflag": False}, {"cartoon": {"color": "lightgray"}})

# Highlight residues in chain B (=p53)
view.setStyle(
    {"chain": "B"},
    {"cartoon": {"color": "red"}})
view.addStyle(
    {"chain": "B", "resi": [19, 23, 26]},
    {"stick": {"color": "red", "radius": 0.25}})

view.zoomTo()
view.setBackgroundColor("white")
view.show()
````

You can see the three interaction p53 residues highlighted in red - these residues are Phe19, Trp23 and Leu26.

## 5. How are protein structures stored on the computer?

Protein structures determined by techniques such as X-ray crystallography, NMR spectroscopy, or cryo-EM are stored on computers as text-based coordinate files that describe the 3D positions of atoms together with experimental and annotation data. The two main formats used by the Protein Data Bank are the legacy PDB format and the more modern mmCIF (Macromolecular Crystallographic Information File) format.

The PDB format is a fixed-column, human-readable text format that has been widely used since the early days of structural biology. Each line (called a “record”) has a specific meaning depending on its keyword. Key record types include:

* HEADER, TITLE, COMPND, SOURCE: general information about the structure, such as the protein name, organism, and experimental context
* EXPDTA: experimental method (e.g., X-ray diffraction)
* REMARK: free-text annotations, including important validation data (e.g., resolution, R-factor)
* CRYST1: crystallographic unit cell parameters and space group
* ATOM / HETATM: the core of the file, listing each atom with fields for atom name, residue name, chain identifier, residue number, 3D coordinates (x, y, z), occupancy, and B-factor
* CONECT: connectivity information (mainly for ligands or nonstandard residues)
* TER / END: delimiters marking chain termination and end of the file

An example ATOM line illustrates the structure:

```ATOM   1123  CA  GLY A 145      12.345  67.890  23.456  1.00 20.50```

Here, the fields specify the atom type (CA), residue (GLY), chain (A), residue number (145), Cartesian coordinates, occupancy, and B-factor.

Despite its simplicity, the PDB format has several limitations, including fixed column widths, restrictions on the number of atoms, and less flexibility for complex metadata. To address these issues, the PDB consortium now uses mmCIF as the primary archival format. mmCIF is a more flexible, dictionary-based format that stores the same information as PDB but in a structured, extensible way using key–value pairs and tables. It can handle large macromolecular complexes, detailed experimental metadata, and validation statistics more consistently.

In practice, both formats are still widely used: PDB files remain popular for quick visualization and compatibility with older software, while mmCIF files are the standard for deposition, validation, and large-scale computational analysis.
